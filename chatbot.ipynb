```python
# ==================================================
# CHATBOT BLOUDAN – CODE COMPLET (ONE CELL)
# ==================================================

# -------- 1) DETECT LANGUAGE (simple & stable)
def detect_lang(text):
    t = text.lower()
    if any(c in text for c in ["أ","ش","ك","ل","ي","ف","ن","ت","م","ر"]):
        return "ar"
    if any(w in t for w in ["ich","bestellen","essen","kann","möchte","bitte","wo"]):
        return "de"
    if any(w in t for w in ["can","order","open","where","menu","please"]):
        return "en"
    return "fr"


# -------- 2) DETECT INTENT
def detect_intent_from_text(text, role="client"):
    t = text.lower()

    if role == "client":
        if any(w in t for w in ["commande","comande","order","bestellen","اطلب","أطلب","طلب"]):
            return "faire_commande"
        if any(w in t for w in ["menu","plat","plats","essen","eat","منيو","أكل","اكل"]):
            return "demande_menu"
        if any(w in t for w in ["recommend","recommande","empfehlen","تقترح","تنصح"]):
            return "recommandation_plat"
        if any(w in t for w in ["hour","open","horaire","adresse","wo","où","وين","فين"]):
            return "infos_restaurant"
        return "salutation"

    return "salutation"


# -------- 3) SIMPLE SENTIMENT ANALYSIS
def analyser_sentiment_texte(text):
    t = text.lower()
    if any(w in t for w in ["merci","good","great","excellent","ممتاز","شكرا","top"]):
        return {"label": "POSITIVE", "score": 0.9}
    if any(w in t for w in ["bad","problem","late","سيء","غالي","مش باهي"]):
        return {"label": "NEGATIVE", "score": 0.8}
    return {"label": "NEUTRAL", "score": 0.6}


# -------- 4) MULTI-LANGUAGE RESPONSES
def repondre_salutation(lang):
    return {
        "fr": "Bonjour 👋 Bienvenue chez Bloudan Restaurant. Comment je peux t’aider ?",
        "en": "Hello 👋 Welcome to Bloudan Restaurant. How can I help you?",
        "de": "Hallo 👋 Willkommen im Bloudan Restaurant. Wie kann ich helfen?",
        "ar": "أهلاً 👋 مرحبا بيك في مطعم بلودان. كيفاش نجم نعاونك؟"
    }[lang]

def repondre_demande_menu(lang):
    return {
        "fr": "📋 Notre menu : Shawarma, Falafel, Kebbé, Hummus.",
        "en": "📋 Our menu: Shawarma, Falafel, Kebbeh, Hummus.",
        "de": "📋 Unser Menü: Shawarma, Falafel, Kibbeh, Hummus.",
        "ar": "📋 المنيو فيه: شاورما، فلافل، كبة، حمص."
    }[lang]

def repondre_recommandation_plat(lang):
    return {
        "fr": "⭐ Je te recommande le Shawarma poulet.",
        "en": "⭐ I recommend the chicken shawarma.",
        "de": "⭐ Ich empfehle das Hähnchen-Shawarma.",
        "ar": "⭐ نقترح عليك شاورما الدجاج."
    }[lang]

def repondre_infos_restaurant(lang):
    return {
        "fr": "📍 Bloudan Restaurant – ouvert de 11h à 23h.",
        "en": "📍 Bloudan Restaurant – open from 11am to 11pm.",
        "de": "📍 Bloudan Restaurant – geöffnet von 11 bis 23 Uhr.",
        "ar": "📍 مطعم بلودان – مفتوح من 11 إلى 23."
    }[lang]

def repondre_faire_commande(lang):
    return {
        "fr": "✅ Bien sûr. Dis-moi le plat et la quantité.",
        "en": "✅ Sure. Tell me the dish and the quantity.",
        "de": "✅ Natürlich. Sag mir das Gericht und die Menge.",
        "ar": "✅ أكيد. قولي شنوّة الطبق والكمية."
    }[lang]


# -------- 5) DISPATCHER
def chatbot_dispatcher(role, intent, lang):
    if role == "client":
        if intent == "salutation":
            return repondre_salutation(lang)
        if intent == "demande_menu":
            return repondre_demande_menu(lang)
        if intent == "recommandation_plat":
            return repondre_recommandation_plat(lang)
        if intent == "infos_restaurant":
            return repondre_infos_restaurant(lang)
        if intent == "faire_commande":
            return repondre_faire_commande(lang)
        return repondre_salutation(lang)
    else:
        return "Admin mode."


# -------- 6) CHAT LOOP
def discuter_avec_chatbot(role="client"):
    print("💬 Chatbot Bloudan – mode", role)
    print("Écris 'exit' pour quitter.\n")

    while True:
        user_text = input("🧑‍💻 Toi : ")

        if user_text.strip().lower() in ["exit", "quit", "stop", "خروج"]:
            print("👋 Fin de la conversation.")
            break

        lang = detect_lang(user_text)
        intent = detect_intent_from_text(user_text, role=role)
        bot_reply = chatbot_dispatcher(role, intent, lang)
        sentiment = analyser_sentiment_texte(user_text)

        print(f"🤖 Bot (intent={intent}, lang={lang}) : {bot_reply}")
        print(f"🧠 Sentiment : {sentiment['label']} (score={sentiment['score']})")
        print("-" * 60)


# -------- 7) RUN CHATBOT
discuter_avec_chatbot(role="client")
```
